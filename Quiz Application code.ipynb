```python
def ask_question(question, options, correct_answer):
    print(question)

    for option in options:
        print(option)

    answer = input("Enter your answer (A/B/C/D): ").upper()

    if answer == correct_answer:
        print("Correct Answer!")
        return 1
    else:
        print("Wrong Answer!")
        print("Correct answer is:", correct_answer)
        return 0


def calculate_score(score, total):
    print(" QUIZ RESULT :")
    print("Your Score:", score, "/", total)

    percentage = (score / total) * 100
    print("Percentage:", percentage, "%")

print("WELCOME TO PYTHON QUIZ")

score = 0

score += ask_question("1. Which function is used to display output in Python?", ["A. input()", "B. print()", "C. scan()", "D. show()"], "B")

score += ask_question("2. Which symbol is used for comments in Python?", ["A. //", "B. /*", "C. #", "D. &"], "C")

score += ask_question("3. Which data type stores True or False?", ["A. int", "B. float", "C. string", "D. bool"],"D")

calculate_score(score, 3)
```
